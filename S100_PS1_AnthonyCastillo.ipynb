# Ejercicio 4 · Buscar relaciones entre variables
**Dificultad: media** · 25 puntos

Hasta aquí miraste variables **una por una**. Ahora las miramos **de a dos**: ¿cuáles se mueven
juntas?

**a)** Calcula la **matriz de correlación** de las 10 variables de la lista `audio`.

**b)** Conviértela en una lista ordenada de pares y encuentra el par **más positivo** y el
**más negativo**. Ojo con dos cosas antes de ordenar:
- la **diagonal** (toda variable correlaciona 1 consigo misma) hay que quitarla;
- la matriz es **simétrica**, así que cada par aparece dos veces: quédate con una sola copia.
Si lo haces bien, tu lista debe tener **45 pares**, no 90.

**c)** Grafica el par **más positivo** con un diagrama de dispersión.
Con 32,833 puntos la nube se vuelve una mancha: usa una muestra y transparencia.

**d)** Calcula ahora la correlación de cada variable de `audio` con `track_popularity`.
¿Cuál es la que más se relaciona con la popularidad?

> 💡 **Pistas.**
> `corr = df[audio].corr()`
> ```python
> pares = corr.unstack()
> pares = pares[pares < 0.999]                                       # fuera la diagonal
> pares = pares[pares.index.get_level_values(0) < pares.index.get_level_values(1)]  # fuera repetidos
> pares = pares.sort_values()
> ```
> El primero y el último de esa lista son los que buscas.
> Para el gráfico: `m = df.sample(4000, random_state=1)` y luego
> `plt.scatter(m["x"], m["y"], s=6, alpha=0.25, color=AZUL, lw=0)`.
> Para (d): `df[audio + ["track_popularity"]].corr()["track_popularity"].sort_values()`.

````python
# a) Matriz de correlación
corr = df[audio].corr()
print("Matriz de correlación:")
print(corr.round(2))
````

````python
# b) Lista de pares ordenada: el más positivo y el más negativo
pares = corr.unstack()
pares = pares[pares < 0.999]                                                   # fuera la diagonal
pares = pares[pares.index.get_level_values(0) < pares.index.get_level_values(1)]  # fuera repetidos
pares = pares.sort_values()

print(f"\nTotal de pares (debe ser 45): {len(pares)}")

print(pares)
````

````python
# c) Diagrama de dispersión del par más positivo
par_mas_positivo = pares.index[-1]

x_var, y_var = par_mas_positivo

m = df.sample(4000, random_state=1)

plt.figure(figsize=(7, 5))
plt.scatter(m[x_var], m[y_var], alpha=0.50)
plt.xlabel(x_var)
plt.ylabel(y_var)
plt.title(f"Relación entre Energia y Volumen")

plt.show()
````

````python
# d) Correlación de cada variable de audio con track_popularity
corr_popularidad = df[audio + ["track_popularity"]].corr()["track_popularity"].sort_values()
corr_popularidad = corr_popularidad.drop("track_popularity")  # quitamos la autocorrelación

print("\nCorrelación de cada variable de audio con track_popularity:")
print(corr_popularidad.round(2))

variable_mas_relacionada = corr_popularidad.abs().idxmax()
print(f"\nVariable más relacionada (en valor absoluto) con la popularidad: "f"{variable_mas_relacionada} (r = {corr_popularidad[variable_mas_relacionada]:.3f})")
````

# Ejercicio 5 · La relación escondida
**Dificultad: media** · 25 puntos

En el Ejercicio 4 calculaste correlaciones sobre **toda la tabla**. Pero en el Ejercicio 3
comprobaste que los géneros son muy distintos entre sí. La pregunta final del problem set es qué

**d)** Estandariza `energy` creando la columna `energy_z`, y encuentra la canción con el

Comprueba además que `energy_z` tiene media 0 y desviación 1, y que su `skew()` es el mismo

> 💡 **Pistas.**
> Correlación por grupo:
> `df.groupby("playlist_genre").apply(lambda g: g["energy"].corr(g["danceability"])).sort_values()`
> El `lambda g:` recibe el sub-DataFrame de cada género.
> Para juntar tablas que comparten el mismo índice (el género): `pd.concat([tabla1, tabla2], axis=1)`.
> Estandarización: `z = (x - x.mean()) / x.std()`.
> `.idxmax()` devuelve el **índice** de la fila con el valor máximo; con `df.loc[idx, [...]]`
> sacas las columnas que quieras de esa fila.
>
> ⚠️ Antes de escribir la conclusión, mira los **signos** de la tabla de (b): ¿todos los géneros
> apuntan hacia el mismo lado?

````python
# a) Correlación global entre energy y danceability
corr_total = df["energy"].corr(df["danceability"])
print("Correlación total energy-danceability:", round(corr_total, 4))
````

````python
# b) La misma correlación dentro de cada género
corr_genero = (df.groupby("playlist_genre").apply(lambda g: g["energy"].corr(g["danceability"]), include_groups=False).sort_values())

#corr_genero.name = "corr_energy_danceability"

print("\ncorr_genero:")
print(corr_genero)

````

````python
# c) Tabla resumen por género
energia_genero = df.groupby("playlist_genre")["energy"].mean()          # del Ejercicio 3
danceability_genero = df.groupby("playlist_genre")["danceability"].mean()
popularidad_genero = df.groupby("playlist_genre")["track_popularity"].mean()

tabla = pd.concat([
    energia_genero.rename("energy_media"),
    danceability_genero.rename("danceability_media"),
    popularidad_genero.rename("track_popularity_media"),
    corr_genero
], axis=1)

print("\nTabla combinada por género:")
print(tabla.round(3))
````

````python
# d) Estandarización de energy y canción más extrema
df["energy_z"] = (df["energy"] - df["energy"].mean()) / df["energy"].std()

idx_max = df["energy_z"].idxmax()
print("\nCanción con energy_z más alto:")
print(df.loc[idx_max, ["track_name", "track_artist", "playlist_genre", "energy", "energy_z"]])

print("\nComprobaciones:")
print("media energy_z:", df["energy_z"].mean())
print("std energy_z:", df["energy_z"].std())
print("skew energy_z:", df["energy_z"].skew())
print("skew energy original:", df["energy"].skew())
````

La relación entre energía y qué tan bailable es una canción cambia según el género: en rock, edm y rap, a más energía la canción tiende a ser menos bailable, pero en latin y r&b ocurre justo lo contrario, y por eso al mezclar todos los géneros en un solo número esa relación parece casi inexistente. Las tendencias opuestas se anulan entre sí en el promedio general.

Esto nos dice que el género de una canción importa mucho más que sus características de audio a la hora de predecir su éxito, y que cualquier receta genérica del tipo sube la energía o hazla más bailable puede funcionar en una dirección para un género y en la contraria para otro. Por eso, si la meta es aumentar la popularidad, tiene más sentido pensar en estrategias específicas por género que en ajustar una sola característica de audio para todo el catálogo.

Eso sí, estos datos muestran patrones y coincidencias entre variables, pero no prueban que cambiar la energía o el nivel de baile de una canción vaya a cambiar su popularidad, podría haber otros factores; como el artista, el marketing o las tendencias del momento, detrás de estos números.
